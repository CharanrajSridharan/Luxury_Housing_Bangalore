```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("Luxury_Housing_Bangalore.csv")
```

```python
# Configuration - column Cleaning. 
```

```python
df['Configuration'] = df['Configuration'].str.strip().str.upper()
```

```python
df["Configuration"]
```

```python
# Micro_Market - column Cleaning. 
```

```python
df['Micro_Market'] = df['Micro_Market'].str.strip().str.title()
```

```python
df['Micro_Market']
```

```python
# Ticket_Price_Cr - column Cleaning. 
```

```python
df['Ticket_Price_Cr'] = df['Ticket_Price_Cr'].replace('[₹,Cr]', '', regex=True)
df['Ticket_Price_Cr'] = df['Ticket_Price_Cr'].astype(float)
df['Ticket_Price_Cr'] = df['Ticket_Price_Cr'] * 10000000
```

```python
df['Ticket_Price_Cr']
```

```python
df
```

```python
#null value handlling 
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.info()
```

```python
# Amenity_Score - column Cleaning. 
```

```python
# Cheacking for outliers because we have null values.
sns.boxplot(x=df['Amenity_Score'])
plt.title("Box Plot for Age")
plt.show()
```

```python
mean_value = df['Amenity_Score'].mean() 
```

```python
df["Amenity_Score"] = df["Amenity_Score"].fillna(mean_value)
```

```python
df.isnull().sum()
```

```python
# Unit_Size_Sqft - column Cleaning. 
```

```python
# fill it with the average size of similar homes. So we do the grouping of places and take the mean and replace them in the null values.

df['Unit_Size_Sqft'] = df['Unit_Size_Sqft'].fillna(
    df.groupby(['Micro_Market', 'Configuration'])['Unit_Size_Sqft'].transform('mean')
)
```

```python

df.isnull().sum()
```

```python
# Ticket_Price_Cr - column Cleaning
```

```python
df['Ticket_Price_Cr'] = df['Ticket_Price_Cr'].fillna(
    df.groupby(['Micro_Market', 'Configuration'])['Ticket_Price_Cr'].transform('mean')
)
```

```python
df.isnull().sum()
```

```python
#Data Type Fixing
```

```python
df.dtypes
```

```python
df['Purchase_Quarter'] = pd.to_datetime(df['Purchase_Quarter'])
```

```python
df['Unit_Size_Sqft'].min()
```

```python
# Cheacking for outliers 
sns.boxplot(x=df['Unit_Size_Sqft'])
plt.title("Box Plot for Age")
plt.show()
```

```python
# since we have invalid value in unit_size we are going to convert it into null value and then replace it with group by mean value 
```

```python
df['Unit_Size_Sqft'] = df['Unit_Size_Sqft'].replace(-1, np.nan)
```

```python
df.isnull().sum()
```

```python

df['Unit_Size_Sqft'] = df['Unit_Size_Sqft'].fillna(
    df.groupby(['Micro_Market', 'Configuration'])['Unit_Size_Sqft'].transform('mean')
)
```

```python
df.isnull().sum()
```

```python
# Feature Engineering
#New column name: price_per_sqft

#Formula (in words): Total property price divided by total area in square feet

#Why useful: It allows fair comparison between properties of different sizes. 
#Instead of comparing total price, we compare cost per unit area, which helps identify expensive vs affordable markets and properties.
```

```python
df['price_per_sqft'] = df['Ticket_Price_Cr'] / df['Unit_Size_Sqft']
```

```python
# New column name: quarter_number

# Formula (in words): Extract the quarter (1, 2, 3, 4) from the purchase_quarter date

# Why useful: Helps analyze seasonal trends, like which quarter has higher sales, pricing trends, or buyer activity.
```

```python
df['quarter_number'] = df['Purchase_Quarter'].dt.quarter
```

```python
df
```

```python
#Categorical Encoding
```

```python
df['NRI_Buyer'] = df['NRI_Buyer'].map({
    'yes': 1,
    'no': 0
})
```

```python
df['NRI_Buyer'].unique()
```
